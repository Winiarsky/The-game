```python
class temp():
    def __init__(self, _id):
        self.id = _id
    def __eq__(self, other):
        return self.id == other.id
    def __hash__(self):
        return hash(self.id)
    def __repr__(self):
        return f"temp({self.id})"
```

```python
a = temp(1)
b = temp(1)
c = temp(2)

print(a == b)
print(a == c)
```
